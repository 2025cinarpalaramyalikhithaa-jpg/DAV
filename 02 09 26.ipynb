```python
import numpy as np
import pandas as pd

arr=np.array([10,20,30,40,50])
print("array:",arr)
print("mean:",arr.mean())
print("reshaped array:",arr.reshape(5,1))
print("sliced (index 1 to 3):",arr[1:4])

data= {'name':['ramya','koush','poorvi','thanu',],'marks':[88,45,46,76],'branches':['cse','isc','ece','aiml']}
df = pd.DataFrame(data)
print(df)

print("\nfirst 2 rows:\n", df.iloc[0:2])
print('\nmarks column:\n', df['marks'])
print("\nrows where marks > 70:\n",df[df['marks'] >70])
                                                
```
